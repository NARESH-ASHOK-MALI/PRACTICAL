```python
from numpy._core.fromnumeric import mean
outliers = []
def detect(df):
  threshold=3
  mean=np.mean(df)
  std=np.std(df)

  for d in df:
    z_score=(d-mean)/std
    if np.abs(z_score)>threshold:
      outliers.append(d)
  return outliers
```

```python
var ="math score"
#var="reading score"
#var="writing score"
```

```python
z_score=detect(df[var])
```

```python
outliers=df[df[var].isin(z_score)]
```
