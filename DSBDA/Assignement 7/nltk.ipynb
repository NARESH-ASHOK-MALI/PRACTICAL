```python
from nltk.corpus import stopwords
import re

text = "How to remove stop words with NLTK library in Python?"

stop_words = set(stopwords.words("english"))

text = re.sub('[^a-zA-Z]', ' ', text)
tokens = word_tokenize(text.lower())

filtered = [w for w in tokens if w not in stop_words]

print(filtered)
```

```python
import math
import pandas as pd

documentA = "Jupiter is the largest Planet"
documentB = "Mars is the fourth planet from the Sun"

# BoW
bagA = documentA.split()
bagB = documentB.split()

# Unique words
uniqueWords = set(bagA).union(set(bagB))

# Word count
numA = dict.fromkeys(uniqueWords, 0)
for word in bagA:
    numA[word] += 1

numB = dict.fromkeys(uniqueWords, 0)
for word in bagB:
    numB[word] += 1

# TF
def computeTF(wordDict, bag):
    tf = {}
    total = len(bag)
    for word, count in wordDict.items():
        tf[word] = count / total
    return tf

tfA = computeTF(numA, bagA)
tfB = computeTF(numB, bagB)

# IDF
def computeIDF(docs):
    N = len(docs)
    idf = dict.fromkeys(docs[0].keys(), 0)

    for doc in docs:
        for word, val in doc.items():
            if val > 0:
                idf[word] += 1

    for word in idf:
        idf[word] = math.log(N / float(idf[word]))

    return idf

idf = computeIDF([numA, numB])

# TF-IDF
def computeTFIDF(tf, idf):
    tfidf = {}
    for word, val in tf.items():
        tfidf[word] = val * idf[word]
    return tfidf

tfidfA = computeTFIDF(tfA, idf)
tfidfB = computeTFIDF(tfB, idf)

df = pd.DataFrame([tfidfA, tfidfB])
print(df)
```
