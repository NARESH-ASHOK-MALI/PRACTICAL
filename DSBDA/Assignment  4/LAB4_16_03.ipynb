```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import numpy as np
from sklearn.metrics import mean_squared_error,r2_score
```

```python
df=pd.read_csv("Boston - Boston.csv")
```

```python
df
```

```python
df.rename({"Unnamed: 0":"a"},axis="columns",inplace=True)
df.drop(["a"],axis=1,inplace=True)
```

```python
rm=df['rm']
medv=df['medv']
```

```python
x_train, x_test, y_train, y_test =train_test_split(rm, medv, test_size = 0.2)
```

```python
x_train.shape
```

```python
y_train.shape
```

```python
x_train=np.array(x_train).reshape(-1,1)
y_train=np.array(y_train).reshape(-1,1)
x_test = np.array(x_test).reshape(-1,1)
y_test = np.array(y_test).reshape(-1,1)
```

```python
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
y_predict=model.predict(x_test)
```

```python
plt.title("predicted results")
plt.xlabel("rm")
plt.ylabel("medv")
plt.scatter(x_train,y_train,color="blue" ,label="Traing Data")
plt.scatter(x_test,y_predict,color="red",label="Actual")
plt.scatter(x_test,y_test,color="yellow"  ,label="Predicted")
```

```python
mse = mean_squared_error(y_test, y_predict)
print(mse)
```

```python
r2=r2_score(y_test, y_predict)
print(r2)
```

```python
rmse=(np.sqrt(mean_squared_error(y_test,y_predict)))
print('Root Mean Squared Error is {}'.format(rmse))
```
